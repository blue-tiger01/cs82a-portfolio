Multiplication the long way

```python
def multiply_for(a, b):
    total = 0
    for i in range(b):
        total += a
    return total


def multiply_while(a, b):
    total = 0
    count = 0
    while count < b:
        total += a
        count += 1
    return total
```

```python
a = 7
b = 6
expected = 42

for_result = multiply_for(a, b)
while_result = multiply_while(a, b)

print(f"For loop:   {a} x {b} = {for_result}")
print(f"While loop: {a} x {b} = {while_result}")
print(f"Known product: {expected}")
print(f"Both match: {for_result == while_result == expected}")
```

```python
print("Check with zero:", multiply_for(7, 0), multiply_while(7, 0))
print("Check 3 x 4:", multiply_for(3, 4), multiply_while(3, 4))
```
